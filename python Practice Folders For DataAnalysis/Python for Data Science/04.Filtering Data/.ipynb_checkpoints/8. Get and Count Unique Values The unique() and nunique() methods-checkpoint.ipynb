```python
# Pandas is a Python library used for working with structured data, such as tables, spreadsheets, and CSV files
# Core Uses of Pandas:
# • Loading and Saving Data
# • Viewing and Inspecting Tables
# • Filtering and Searching
# • Data Cleaning
# • Math and Grouping
import pandas as pd
```

```python
df_laptops = pd.read_csv('0.2_laptop_price.csv', encoding='latin1')
```

```python
df_laptops.head(3)
```

# unique()
Return unique values of Series object. (Uniques are returned in order of appearance.)

```python
# get unique elements in the "Company" column
df_laptops['Company'].unique()
```

```python
# get unique elements in the "Inches" column
df_laptops['Inches'].unique()
```

```python
# get the length of unique elements
len(df_laptops['Inches'].unique())
```

```python
# get the length of unique elements
len(df_laptops['Company'].unique())
```

# nunique()
Return number of unique elements in the object (Excludes NA values by default)

```python
# get number of unique elements in the "Company" column
df_laptops['Company'].nunique()
```

```python
# get number of unique elements in the "Inches" column
df_laptops['Inches'].nunique()
```

```python
# When there are missing values (NaN): .nunique() ignores NaN by default.
# If you want .nunique() to count NaN just like len(.unique()) does, pass dropna=False:
df_laptops['Inches'].nunique(dropna=False)
```
