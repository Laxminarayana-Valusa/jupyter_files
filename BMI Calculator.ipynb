```python
name = input("Enter your name:")

wt = int(input("Enter you weight in pound:")) 
ht =  int(input("Enter you height in inch:")) 

BMI = (wt*703)/(ht**2)

print('Your BMI is:', BMI)

if BMI>0:
    if BMI<18.5:
        print(name, "you are Underweight: ", BMI, "you need Minimal Excercise")
    elif BMI<=24.9:
        print(name, "you are Normal weight: ", BMI, "you need Minimal Excercise")
    elif BMI<=29.9:
        print(name, "you are overweight: ", BMI, "you need to Increase your Excercise")
    elif BMI<=34.9:
        print(name, "you are obese: ", BMI, "you need to do High Excercise")
    elif BMI<=39.9:
        print(name, "you are Severely obese: ", BMI, "you need to do Very high Excercise")
    else:
        print(name, "you are over Morbidly obese: ", BMI, "you need to do Very high Excercise")
else:
    print("Enter valid input")
```
