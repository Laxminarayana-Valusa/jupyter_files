```python
# Import libraries for web scraping, timing, and sending emails

# BeautifulSoup is used to parse HTML and extract specific data from web pages.
# bs4 is BeautifulSoup version 4 package in python
# BeautifulSoup is a library from bs4
from bs4 import BeautifulSoup

# requests allows you to send HTTP requests to websites to download their HTML content.
import requests

# time handles time-related tasks, like pausing your code (time.sleep) between loops.
import time

# datetime lets you fetch, track, and format current dates and timestamps.
import datetime

# smtplib stands for "Simple Mail Transfer Protocol library"/ tool 
# It is used to connect to an email server and automatically send emails from your script.
import smtplib

```

```python
# url is variable that stored website link in the form of string format
url = 'https://www.amazon.in/Lux-Cozi-Maroon-Cotton-Tshirt/dp/B0D395SFC2/ref=sr_1_14?_encoding=UTF8&content-id=amzn1.sym.f01fdba1-e898-409d-9b1d-3d6b8106a360&dib=eyJ2IjoiMSJ9.dsV8Y4GFcB6ivYjn_pD5JKcik2ryOJzJdYhcbd8KR0Li5TX5Wd2ufbWf6D6tV6d3-o0vo_y5KNTOBpjecYxMWs3RJa3M-R-jypy0gTSFJ3BcoSoLQvod_8IQ0zyKQFjjHscKBS4s1UyxStOzQlVBBjUooivyxfDAoWA0uSEGToCIyNon8drq0Sce8DAuFFcRQwfHrHLGLtk4YkoLlwsH8CvkHJwKLiTetVXWjv7_zHAO0lNJSI87mGCxOzoKv1y6xcvzhDDZNeP_J9wTBtJ-a2ZsbVKSy_2sp5ZhwdIU1Mg.6_pe1QQeG5ca91mnSr59r2EJM2Bdy5DcgkLk3nhh-oU&dib_tag=se&pd_rd_r=cf29a906-98a7-493b-aa3d-3fbd26084ae1&pd_rd_w=yxB7E&pd_rd_wg=kVg6g&qid=1791705578&refinements=p_85%3A10440599031%2Cp_36%3A-40000%2Cp_28%3ATees%7CTshirt%7Cpolos%7Ctshirt%7Ctshirts&rps=1&s=apparel&sr=1-14&th=1&psc=1'
# Create a dictionary of headers to trick websites into thinking our code is a real human.

# headers = {
#     # Pretend to be a real Google Chrome browser running on Windows 10.
#     "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/78.0.3904.108 Safari/537.36", 
#     # Say we can accept fast, zipped/compressed data.
#     "Accept-Encoding": "gzip, deflate", 
#     # Ask the server to send us standard webpage text (HTML).
#     "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,*/*;q=0.8", 
#     # "Do Not Track" - Ask the website not to track our behavior.
#     "DNT": "1",
#     # Disconnect from the website as soon as the download finishes.
#     "Connection": "close", 
#     # Always ask for the safe, encrypted (HTTPS) version of the page.
#     "Upgrade-Insecure-Requests": "1"
# }
headers = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/78.0.3904.108 Safari/537.36", "Accept-Encoding":"gzip, deflate", "Accept":"text/html,application/xhtml+xml,application/xml;q=0.9,*/*;q=0.8", "DNT":"1","Connection":"close", "Upgrade-Insecure-Requests":"1"}


# Send an HTTP GET request to download the webpage data from the specified URL.
# - requests.get(url, ...) contacts the website's server to ask for its content.
# - headers = headers passes our custom user-agent browser settings to avoid being blocked.
# - The resulting website data, status codes, and raw HTML are saved into the 'page' variable.
page = requests.get(url, headers = headers)


# Feed the raw HTML content from the downloaded website into the BeautifulSoup parser.
# - page.content extracts the raw, unfiltered bytes/HTML text from the web server response.
# - 'html.parser' specifies Python's built-in tool used to break down and read HTML syntax.
# - 'soup' stores the final, organized, and searchable tree structure of the entire webpage.
soup1 = BeautifulSoup(page.content, 'html.parser')
# print(soup)


# The basic difference comes down to data type:
# - page.content = Raw bytes (binary data). Best for BeautifulSoup and images.
# - page.text    = Plain text (a Python string). Best for reading with human eyes.



# .prettify() takes messy, unorganized HTML code and formats it beautifully.
# - It cleans up the layout by adding proper line breaks and neat indentations.
# - This line runs .prettify() on 'soup1', then re-feeds that clean text into 
#   BeautifulSoup again to save a perfectly structured tree inside 'soup2'.
soup2 = BeautifulSoup(soup1.prettify(), "html.parser")

title = soup2.find(id='title').get_text()
# expected_result = Lux Cozi Men Cotton Half Sleeves Round Neck Regular Fit Casual T-Shirt
print(title)

price = soup2.find(class_='a-size-base').get_text()
# expected_result = 249
print(price)

```

```python
# cleanUp the data little bit

title = title.strip()

# price = price.strip()[1:]
price = price.strip()
print(title)
print(price)
```

```python
# Create a Timestamp for your output to track when data was collected

import datetime

today = datetime.date.today()

print(today)
```

```python
# Create a CSV file and write both column headers and your scraped data rows into it.


# Import Python's built-in module for reading and writing comma-separated values (CSV) files.

# import csv allows you to use Python's built-in Comma-Separated Values module.
# It gives your script the tools needed to easily read data out of CSV files, 
# or generate and format new tables into standard .csv files that Excel can open.

import csv 

# Define the top row column headers as a list of text strings.
header = ['Title', 'Price', 'Date']

# Wrap your scraped variables (title, price, today's date) inside a single row list.
data = [title, price, today]

# data1 = ['Apple', 'banana', 'mango']


# Open (or create) a file named 'AmazonWebScraperDataset.csv' in write ('w') mode.
# - newline='' prevents Python from accidentally adding extra blank rows between lines.
# - encoding='UTF8' ensures international text or symbols (like currency) save correctly.
# - 'as f' creates a temporary file variable named 'f' that automatically closes when done.
with open('AmazonWebScraperDataset.csv', 'w', newline='', encoding='UTF8') as f:
    
    # Create a specialized writing object that translates list data into a CSV string structure.
    writer = csv.writer(f)
    
    # Write the column labels ('Title', 'Price', 'Date') as the very first row.
    writer.writerow(header)
    
    # Write your actual extracted data entry directly beneath the headers.
    writer.writerow(data)
   # writer.writerow(data1)
```

```python
# Load the powerful pandas library into your script and nickname it 'pd' to save typing.
import pandas as pd

# Open the CSV file from your computer's C drive and save its entire table into a variable named 'df'.
# The 'r' before the file path ensures Windows folder slashes (\) do not cause bugs.
df = pd.read_csv(r"C:\Users\valus\python Practice Folders For DataAnalysis\AmazonWebScraperDataset.csv")

# Display the entire table cleanly on your screen so you can inspect your columns and rows.
print(df)
```

```python
# now we are appending the data to the csv

# The mode 'a+' stands for "Append and Read".
# - 'a' (Append): Opens the file to add new data directly to the very bottom. 
#   It never deletes your existing data. If the file doesn't exist, it creates a new one.
# - '+' (Plus): Unlocks extra power, allowing your script to both read from and write to the file.
with open('AmazonWebScraperDataset.csv', 'a+', newline='', encoding='UTF8') as f:
        
    writer = csv.writer(f)
    writer.writerow(data)
```
