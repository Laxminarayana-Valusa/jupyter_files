```python
import os, shutil
```

```python
path = r"C:/Desktop/data analyst/python"
```

```python
csv_file = os.path.join(path,"csv file")
os.path.exists(csv_file)
```

```python
os.path.exists(path+'/csv file')
```

```python
os.listdir(path)
```

```python
file_names = os.listdir(path)
for file in file_names:
    # to skip the hidden window file
    if file == 'desktop.ini':
        continue
    print("Processing real file:", file)
```

```python
folder_names = ['xlsx files', 'image files', 'text files', 'csv files']

# A cleaner way to Loop through items directly
for folder in folder_names:

# To Just print folder name if not exists
    if not os.path.exists(path+'/'+folder):
        print("Creating:"+ path+'/'+folder)
        
# To create file folder if really not existed
        os.makedirs(path+'/'+folder)

# To move files from source folder to created dist_folders
for file in file_names:
    if file == 'desktop.ini' or file in folder_names:
        continue
    if ".xlsx" in file and not os.path.exists(path+"/xlsx files/"+file):
         shutil.move(path+'/'+file, path+"/xlsx files/"+file)
    elif ".png" in file and not os.path.exists(path+"/image files/"+file):
         shutil.move(path+'/'+file, path+"/image files/"+file) 
    elif ".jpg" in file and not os.path.exists(path+"/image files/"+file):
          shutil.move(path+'/'+file, path+"/image files/"+file)
    elif ".txt" in file and not os.path.exists(path+"/text files/"+file):
          shutil.move(path+'/'+file, path+"/text files/"+file)
    elif ".csv" in file and not os.path.exists(path+"/csv files/"+file):
          shutil.move(path+'/'+file, path+"/csv files/"+file)
```
