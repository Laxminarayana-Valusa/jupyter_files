```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Desktop\data analyst\python\Excercise Files\world_population.csv")
df
```

```python
# ==============================================================================
# CONFIGURING PANDAS DISPLAY FORMAT FOR FLOATING-POINT NUMBERS
# ==============================================================================

# pd.set_option() is a global pandas function used to adjust library-wide settings.
# Here, we target "display.float_format" to alter how decimals are printed to the console/notebook.
# 
# We assign a 'lambda' function—a small, anonymous, inline function—to this option.
# - 'lambda x:' declares the function and establishes 'x' as the incoming float value from the DataFrame.
# - '"%.2f" % x' is a classic Python string formatting technique (C-style formatting).
#   - '%' starts the format specifier.
#   - '.' indicates that we want to control the precision after the decimal point.
#   - '2' strictly limits the output to exactly two decimal places.
#   - 'f' treats the input 'x' specifically as a floating-point number.
#   - '% x' passes the actual data value into the string template.
#
# IMPORTANT: This is purely a visual presentation layer configuration. 
# It does NOT truncate, round, or alter the underlying precision of your data in memory.
# All mathematical calculations will still use the full, original precision of the numbers.

pd.set_option("display.float_format", lambda x: "%.2f" % x)
```

```python
# Print a quick, helpful summary of the entire DataFrame.
# It instantly shows you:
# 1. How many total rows and columns you have.
# 2. The name of every single column.
# 3. How many missing or blank values are in each column.
# 4. The data type of each column (like text, numbers, or dates).
# 5. How much computer memory (RAM) the data is using.
df.info()
```

```python
# Generate a statistical summary of the DataFrame.
# By default, it automatically analyzes all numeric columns and shows:
# 1. count: The number of non-empty values.
# 2. mean: The mathematical average.
# 3. std: The standard deviation (how spread out the numbers are).
# 4. min / max: The lowest and highest values in the column.
# 5. 25% / 50% / 75%: The percentiles (50% is the median/middle value).
df.describe()
```

```python
# Count how many missing (NaN/blank) values are in each column.
# - df.isnull() checks every cell and turns it into True (if empty) or False (if full).
# - .sum() adds them up per column (treating True as 1 and False as 0).
# The output gives you a clean list of columns and their total missing row counts.
df.isnull().sum()
```

```python
# Count how many unique (distinct) values are in each column.
# - It looks at every column and counts only the different, unique entries.
# - It automatically ignores missing or blank (NaN) values.
# This helps you instantly see if a column contains categories (like 'Gender') 
# or completely unique IDs (like 'User_ID').
df.nunique()
```

```python
# Find and display the top 3 rows with the highest "World Population Percentage".
# - df.sort_values(...) rearranges the rows of the DataFrame based on a specific column.
# - by = "World Population Percentage" specifies the column to look at for sorting.
# - ascending = False ensures the data is sorted from largest to smallest (descending order).
# - .head(3) keeps only the first 3 rows of that sorted result.
df.sort_values(by = "World Population Percentage" , ascending = False).head(3)
```

```python
# Calculate how numeric columns relate to each other (correlation matrix).
# - df.corr() measures the linear strength and direction between pairs of numbers.
# - numeric_only = True forces pandas to only look at columns with numbers,
#   safely ignoring columns containing text, dates, or categories.
# The resulting scores range from -1 (opposite) to +1 (perfect match), where 0 means no relation.
df.corr(numeric_only = True)
```

```python
# Create and display a color-coded heatmap of the numeric correlations.
# - sns.heatmap(...) draws a grid of colored squares based on the correlation matrix.
# - annot = True writes the actual correlation numbers inside each square.
# - plt.rcParams['figure.figsize'] = (20,7) sets the window width to 20 inches and height to 7 inches.
# - plt.show() displays the finalized plot on your screen.
sns.heatmap(df.corr(numeric_only = True), annot = True)
plt.rcParams['figure.figsize'] = (20,7)

plt.show()
```

```python
# groups all rows based on Continent
# finds individual Continent mean values
# sorts_values orders the values
# selects column by 2022 population
# ascending False means high to low
df.groupby('Continent').mean(numeric_only = True).sort_values(by="2022 Population",ascending=False)
```

```python
# scans the Continent column using string method and
# returns True/False inside df[....]
# drops False values rows completly
# gives only True value rows 
df[df['Continent'].str.contains('Oceania')]
```

```python
# groups all Countries by Continents
# in double sq bracket selected columns comes in result
# mean method calculates means of each continent in each column
# sort_values method orders data in columns
# by is arg/parameters based on 2022 population
# ascending is arg/parameters
# False means reverse case 

df2 = df.groupby('Continent')[['1970 Population',
       '1980 Population', '1990 Population', '2000 Population',
       '2010 Population', '2015 Population', '2020 Population',
       '2022 Population']].mean().sort_values(by="2022 Population",ascending=False)
df2
```
