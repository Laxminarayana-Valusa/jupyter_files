```python
from bs4 import BeautifulSoup
import requests
import pandas as pd
```

```python
# Custom identifying header
headers = {
    'User-Agent': 'MyDataScraperBot/1.0 (contact@example.com)'
}

url = 'https://en.wikipedia.org/wiki/List_of_largest_companies_in_the_United_States_by_revenue'
requests.get(url)
page = requests.get(url, headers = headers)
```

```python
soup = BeautifulSoup(page.text, 'html')
```

```python
All_table = soup.find_all('table')
table = All_table[2]
table
```

```python
world_titles = table.find_all('th')
world_titles
```

```python
world_table_titles = [titles.text.strip() for titles in world_titles]
world_table_titles
```

```python
df = pd.DataFrame(columns = world_table_titles)
```

```python
df
```

```python
column_data = table.find_all('tr')
```

```python
for row in column_data[1:]:
    row_data = row.find_all('td')
    indivisual_row_data = [data.text.strip() for data in row_data]

    length = len(df)
    df.loc[length] = indivisual_row_data
```

```python
df.to_csv('C:\Desktop\data analyst\Pandas\ListByCompaniesProfit3.csv', index = False)
```
