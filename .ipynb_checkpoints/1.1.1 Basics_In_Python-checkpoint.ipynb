7) Case change


```python
x = "hElLo wORld"
print(x.upper(), end="\n\n")
print(x.lower() + "\n")

# title case first letter of each word beconmes capital
print(x.title())
print()  # Creates a line break


# capitalize first letter of entire string beconmes capital
print(x.capitalize(), end="\n\n")


# count letter in entire string or word
print(x.count('l'), end="\n\n")

# replace char in entire string or word, case sensitive
print(x.replace('l', 'z'))

# replace capital L with Z
print(x.replace('L', 'Z'))

```

12) Removing an elements

```python
countries = ['United States', 'India', 'China', 'Brazil']
print(countries, end='\n\n')

print("remove listed name from lists")
countries.remove('United States')
print(countries, end='\n\n')
```

```python
print("removes listed index value from list")
countries.pop(1)
print(countries)
```

```python
print("print all countries list", end='\n\n')
countries = ['United States', 'India', 'China', 'Brazil']
print(countries, end='\n\n')
```

```python
# removes listed index value from list
countries = ['United States', 'India', 'China', 'Brazil']
print(countries)
countries.pop(0)
print(countries)
```

```python
# removes listed index value from list using negative value
countries = ['United States', 'India', 'China', 'Brazil']
print(countries)
countries.pop(-1)
print(countries)
```

```python
countries = ['United States', 'India', 'China', 'Brazil']
print(countries)
del countries[0]
print(countries)
```

19) If Statement

```python
# crtl + /  for all selected text commenting in multilines
# if <condition>:
#     <code>
# elif <condition>:
#     <code>
# ...
# else:
#     <code>  
# 


age = 20
if age>=20:
    print("You're an adult!")
elif age>=13:
    print("You're a teenager")
else:
    print("You're a kid")
```

18) For loop

```python
# for <variable> in <iterable>:
#     <code>


# for <variable> in <list>
#        <code>
```

```python
countries
```

```python
for country in countries:
    print(country)
```

```python
# number of the loop
for i, country in enumerate(countries):
    print(i)
    print(country)
```

```python
my_data = {'name': 'Apple', 'age': 26}
my_data
```

```python
for key, value in my_data.items():
    print(key)
    print(value)
```

```python
for key, value in my_data.items():
    print(key)
    print(value)
```

 19) Function

```python
# def function(<params>):
#     <code>
#     return <data>
```

```python
# def = define, create a new, custom function
def sum_values(a,b):
    x = a + b
    return x
```

```python
print(sum_values(10, 20))
print(sum_values(10, 90))
```
