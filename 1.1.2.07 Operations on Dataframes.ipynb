```python
import pandas as pd
```

```python
# reading the csv file
df_exams = pd.read_csv('0.1_data_StudentsPerformance.csv')
df_exams.head()
```

# 1 Math Operations

1.1 Operations in columns

```python
# select a column and calculate total sum
df_exams['math score'].sum()
```

```python
# count, mean, std, max, and min
count = df_exams['math score'].count()
mean = df_exams['math score'].mean()
std = df_exams['math score'].std()
maxs = df_exams['math score'].max()
mins = df_exams['math score'].min()

print(count)
print(mean)
print(std)
print(maxs)
print(mins)
```

```python
# easier calculation with .describe()
df_exams.describe()
```

1.2 Operations in rows

```python
# calculating the sum in a row
df_exams['math score'] + df_exams['reading score'] + df_exams['writing score']   
```

```python
# calculating the avg score and assigning the result to a new column
df_exams['avg'] = (df_exams['math score'] + df_exams['reading score'])/3
```
