```python
# read, clean, and analyze spreadsheet-style data using a short, standard alias
import pandas as pd
```

```python
# reading the csv file
df_exams = pd.read_csv('0.1_data_StudentsPerformance.csv')
df_exams
```

# 1 Create Index

```python
#  NumPy library for fast math and arrays,
import numpy as np

# random  loads Python's built-in module for generating basic random numbers and choices
import random
```

```python
# Creating non repetitive values for the index
new_index = np.arange(0, 1000)
```

```python
# shuffling the indexes
random.shuffle(new_index)
```

```python
# create a new column with new index
df_exams['new_index'] = new_index
df_exams
```

# 2 Set Index

```python
# set new_index column as index
df_exams.set_index('new_index')
```

```python
# set new_index column as index
df_exams.set_index('new_index', inplace = True)
```

# 3 Sort Index

```python
# sort dataframe by new index
df_exams.sort_index()
```

```python
# sort desc by index
df_exams.sort_index(ascending = False)
```

```python
# sort desc by index
df_exams.sort_index(ascending = False, inplace=True)
```
