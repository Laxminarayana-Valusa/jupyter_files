```python
#BeautifulSoup is tool from bs4 library in python 

#BeautifulSoup tool is used for web scraping to extract data out of HTML and XML files

"""• BeautifulSoup is a shovel. It is only used to dig data out of the messy ground (web pages).
     Once it hands you the data, its job is finished. """
     
#   • pandas is a factory. It takes the raw data, organizes it into neat tables, cleans it up, and performs calculations.


# pandas is a library (specifically, a data analysis and manipulation library).



# requests is a library for making HTTP requests.

# • BeautifulSoup takes that raw HTML and finds/extracts specific text or links from it.
# • pandas takes that extracted text and organizes it into a clean data table.


from bs4 import BeautifulSoup
import requests

import pandas as pd # it imports pandas library from python
```

```python
url = "https://www.google.com/"  # url here is a variable to store string of website link
requests.get(url)                # requests is a python library for making HTTP requests


"""
   • 100 Continue | 101 Switching Protocols | 102 Processing | 103 Early Hints
• 200 OK (Success) | 201 Created | 202 Accepted | 204 No Content | 206 Partial Content
• 300 Multiple Choices | 301 Moved Permanently | 302 Found (Temp Redirect) | 304 Not Modified | 307 Temp Redirect | 308 Permanent Redirect
• 400 Bad Request | 401 Unauthorized | 402 Payment Required | 403 Forbidden | 404 Not Found | 405 Method Not Allowed | 408 Request Timeout | 409 Conflict | 429 Too Many Requests
• 500 Internal Server Error | 501 Not Implemented | 502 Bad Gateway | 503 Service Unavailable | 504 Gateway Timeout | 505 HTTP Version Not Supported

"""

page = requests.get(url) # page stores a response Object
page.url # url link
page.headers # website's security info, file size, or server type
page.content # used if you are downloading images, videos, or PDFs
page.text    # Stores the raw HTML code of the page as text (this is what you give to BeautifulSoup)
page.status_code # Stores the HTTP code number
```

```python
soup = BeautifulSoup(page.text, 'html.parser') #  it stores and outputs the entire parsed HTML document tree as a massive string
print(soup)  # soup is a variable
```

```python
# Custom identifying header
headers = {
    'User-Agent': 'MyDataScraperBot/1.0 (contact@example.com)'
}


page = requests.get("https://en.wikipedia.org/wiki/List_of_highest-grossing_films", headers = headers)
page
```

```python
soup = BeautifulSoup(page.text, 'html.parser')
```

```python
print(soup)
```

```python
all_tables = soup.find_all('table')
table1 = all_tables[0]
table1
```

```python
world_titles = table1.find('tr')
world_titles
```

```python
world_table_titles = [titles.text.strip() for titles in world_titles]
world_table_titles
```

```python
df = pd.DataFrame(columns = world_table_titles)
```

```python
df
```

```python
column_data = table1.find_all('tr')
column_data
```

```python
for row in column_data[1:]:
    row_data = row.find_all('td')
    indivisual_row_data = [data.text.strip() for data in row_data]

    length = len(df)
    df.loc[length] = indivisual_row_data
    
```
