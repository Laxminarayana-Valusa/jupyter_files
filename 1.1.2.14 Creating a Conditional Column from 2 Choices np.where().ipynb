# Creating a Conditional Column from 2 Choices: np.where()

```python
#  used for fast mathematical calculations and working with multi-dimensional arrays
# Core Uses of NumPy:
# • Fast Arrays
# • Vectorized Math
# • Linear Algebra & Statistics
# • Foundation for Pandas
import numpy as np
```

```python
# create an array based on price tiers
np.where(df_laptops['Price_euros']>2000, 'Expensive', 'Cheap')
```

```python
# set it to a new column
df_laptops['price_tiers']=np.where(df_laptops['Price_euros']>2000, 'Expensive', 'Cheap')
```

```python
df_laptops.head()
```

```python
# count values in price_tier column
df_laptops.value_counts('price_tiers')
```
